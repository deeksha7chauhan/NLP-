# **Steps:- **

2. 1-hot rep--index from index of sentences

vocab_size = 10000, dim =10

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = ['the glass of milk',
        'the glass of juice',
        'the cup of tea',
        'I am a good boy',
        'I am a good developer',
        'understand the meaning of words',
        'your videos are good']
```

```python
#Vocab size
vocab_size = 10000
```

### 1-hot representaion - gives index in vocab

```python
onehot_repr = []
for words in sent:
    onehot_repr.append(one_hot(words, vocab_size))
print(onehot_repr)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
```

```python
sent_len = 8
```

```python
embedding_doc = pad_sequences(onehot_repr, padding = 'pre', maxlen = sent_len)
embedding_doc
```
